```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestRegressor
```

Load Data

```python
train = pd.read_csv("E:\\Sample_arvyax_reflective_dataset.xlsx - Dataset_120.csv")
test = pd.read_csv("E:\\arvyax_test_inputs_120.xlsx - Sheet1.csv")
```

Data Cleaning

```python
for df in [train, test]:
    df["journal_text"] = df["journal_text"].fillna("")
    
    for col in ["sleep_hours","energy_level","stress_level","duration_min"]:
        if col in df.columns:
            df[col] = df[col].fillna(train[col].mean())
    
    for col in ["ambience_type","time_of_day","previous_day_mood","face_emotion_hint","reflection_quality"]:
        if col in df.columns:
            df[col] = df[col].fillna("unknown")
```

Feature Engineering

```python
text_feature = "journal_text"
num_features = ["sleep_hours", "energy_level", "stress_level", "duration_min"]
cat_features = ["ambience_type", "time_of_day", "previous_day_mood", "face_emotion_hint"]

preprocessor = ColumnTransformer([
    ("text", TfidfVectorizer(max_features=3000), text_feature),
    ("num", StandardScaler(), num_features),
    ("cat", OneHotEncoder(handle_unknown='ignore'), cat_features)
])
```

Train Emotion Model (Classification)

```python
# Split features and target
X = train.drop(["emotional_state","intensity"], axis=1)

y_state = train["emotional_state"]
y_intensity = train["intensity"]
model_state.fit(X, y_state)
model_intensity.fit(X, y_intensity)
model_state = Pipeline([
    ("prep", preprocessor),
    ("clf", LogisticRegression(max_iter=1000))
])

model_intensity = Pipeline([
    ("prep", preprocessor),
    ("reg", RandomForestRegressor(n_estimators=50, random_state=42))
])
```

Predictions

```python
# TRAIN
model_state.fit(X, y_state)
model_intensity.fit(X, y_intensity)
import pickle

pickle.dump(model_state, open("model_state.pkl", "wb"))

print("Training Done ✅")

# PREDICT
pred_state = model_state.predict(test)
pred_intensity = model_intensity.predict(test)

print("Prediction Done ✅")
```

Confidence

```python
probs = model_state.predict_proba(test)
confidence = probs.max(axis=1)
uncertain_flag = (confidence < 0.6).astype(int)
```

Decision Engine (CORE LOGIC)

```python
def decide_action(state, intensity, stress, energy, time):
    
    if stress >= 8:
        return "box_breathing", "now"
    
    if state in ["sad","lonely"]:
        return "journaling", "now"
    
    if state == "anxious":
        return "grounding", "now"
    
    if energy >= 7:
        return "deep_work", "now"
    
    if energy <= 3:
        return "rest", "later_today"
    
    return "light_planning", "within_15_min"
def generate_message(state, intensity, action):
    
    if state == "sad":
        return "It's okay to feel low. Try writing your thoughts—it can help."
    
    if state == "anxious":
        return "Take a deep breath. Focus on the present moment."
    
    if state == "happy":
        return "Great to hear! Keep going like this."
    
    if state == "angry":
        return "Pause for a moment. Breathe slowly and relax."
    
    return "Take things step by step. You're doing fine."
```

Decision Apply

```python
actions = []
times = []

for i in range(len(test)):
    action, when = decide_action(
        pred_state[i],
        pred_intensity[i],
        test["stress_level"].iloc[i],
        test["energy_level"].iloc[i],
        test["time_of_day"].iloc[i]
    )
    actions.append(action)
    times.append(when)
messages = []

for i in range(len(test)):
    msg = generate_message(
        pred_state[i],
        pred_intensity[i],
        actions[i]
    )
    messages.append(msg)
```

Final Output

```python
actions = []
times = []

for i in range(len(test)):
    action, when = decide_action(
        pred_state[i],
        pred_intensity[i],
        test["stress_level"].iloc[i],
        test["energy_level"].iloc[i],
        test["time_of_day"].iloc[i]
    )
    actions.append(action)
    times.append(when)
```

Save

```python
output = pd.DataFrame({
    "id": test["id"],
    "predicted_state": pred_state,
    "predicted_intensity": np.round(pred_intensity,2),
    "confidence": np.round(confidence,2),
    "uncertain_flag": uncertain_flag,
    "what_to_do": actions,
    "when_to_do": times,
    "support_message": messages 
})
output.to_csv("predictions.csv", index=False)
print(output.head())
```
